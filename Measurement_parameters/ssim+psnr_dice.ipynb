```python
from math import log10, sqrt
import cv2
import numpy as np
import os

def PSNR(original, compressed):
    mse = np.mean((original - compressed) ** 2)
    if mse == 0:  # No error means infinite PSNR; we return 100 dB as a placeholder.
        return 100
    max_pixel = 255.0
    psnr = 20 * log10(max_pixel / sqrt(mse))
    return psnr

if __name__ == "__main__":
    # Directories where your images are saved
    real_dir = '/content/drive/MyDrive/eval/actual'
    fake_dir = '/content/drive/MyDrive/eval/fake_dice'

    psnr_values = []  # List to store PSNR values

    # Loop over patient IDs 400 to 450 (inclusive)
    for patient in range(400, 451):
        # Construct file paths using zero-padded numbering
        real_filename = f"patient{patient:04d}_actual.png"
        fake_filename = f"patient{patient:04d}_fake.png"
        real_path = os.path.join(real_dir, real_filename)
        fake_path = os.path.join(fake_dir, fake_filename)

        # Read images
        original = cv2.imread(real_path)
        compressed = cv2.imread(fake_path, 1)  # Color image

        # Check if images were loaded successfully
        if original is None or compressed is None:
            print(f"Warning: Could not load images for patient {patient}. Skipping this pair.")
            continue

        # If image sizes do not match, resize the compressed image to match the original
        if original.shape != compressed.shape:
            # cv2.resize expects size as (width, height)
            compressed = cv2.resize(compressed, (original.shape[1], original.shape[0]))

        # Compute PSNR for this pair
        psnr_value = PSNR(original, compressed)
        print(f"Patient {patient}: PSNR value is {psnr_value:.2f} dB")
        psnr_values.append(psnr_value)

    # Compute and print the average PSNR if any valid pairs were found
    if psnr_values:
        avg_psnr = np.mean(psnr_values)
        print(f"\nAverage PSNR over all valid image pairs: {avg_psnr:.2f} dB")
    else:
        print("No valid image pairs found.")
```

```python
from skimage.metrics import structural_similarity as ssim
import cv2
import numpy as np
import os

def calculate_SSIM(original, compressed):
    # Convert images to grayscale if they are color images
    if len(original.shape) == 3:
        original_gray = cv2.cvtColor(original, cv2.COLOR_BGR2GRAY)
        compressed_gray = cv2.cvtColor(compressed, cv2.COLOR_BGR2GRAY)
    else:
        original_gray = original
        compressed_gray = compressed

    # Calculate SSIM
    ssim_value = ssim(original_gray, compressed_gray)
    return ssim_value

# Directories where your images are saved
real_dir = '/content/drive/MyDrive/eval/actual'
fake_dir = '/content/drive/MyDrive/eval/fake_dice'

ssim_values = []  # List to store SSIM values

# Loop over patient IDs 400 to 450 (inclusive)
for patient in range(400, 451):
    # Construct file paths using zero-padded numbering
    real_filename = f"patient{patient:04d}_actual.png"
    fake_filename = f"patient{patient:04d}_fake.png"
    real_path = os.path.join(real_dir, real_filename)
    fake_path = os.path.join(fake_dir, fake_filename)

    # Read images
    original = cv2.imread(real_path)
    compressed = cv2.imread(fake_path, 1)  # Color image

    # Check if images were loaded successfully
    if original is None or compressed is None:
        print(f"Warning: Could not load images for patient {patient}. Skipping this pair.")
        continue

    # If image sizes do not match, resize the compressed image to match the original
    if original.shape != compressed.shape:
        # cv2.resize expects size as (width, height)
        compressed = cv2.resize(compressed, (original.shape[1], original.shape[0]))

    # Compute SSIM for this pair
    ssim_value = calculate_SSIM(original, compressed)
    print(f"Patient {patient}: SSIM value is {ssim_value:.4f}")
    ssim_values.append(ssim_value)

# Compute and print the average SSIM if any valid pairs were found
if ssim_values:
    avg_ssim = np.mean(ssim_values)
    print(f"\nAverage SSIM over all valid image pairs: {avg_ssim:.4f}")
else:
    print("No valid image pairs found.")
```
